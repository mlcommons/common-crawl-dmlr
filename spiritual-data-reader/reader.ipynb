# Spiritual Data Reader

```python
import os
from datasets import load_from_disk
from random import shuffle

# Get all paths to test datasets
dirs = [os.path.join('final_data', f, 'test') for f in os.listdir('final_data') if os.path.isdir(os.path.join('final_data', f))]
print(dirs)

tes_set = []
        
for dir in dirs:
    ds = load_from_disk(dir)
    for i in range(len(ds)):
        tes_set.append("{0}\t{1}".format(dir.split(os.sep)[1], ds[i]['text']))

shuffle(tes_set)
with open('test_set.txt', 'w') as f:
    for line in tes_set:
        f.write(line + '\n')
# ds = load_from_disk('final_data/aba_Latn/test')
# for i in range(10):
#     print(len(ds))
```
